```python
from dotenv import load_dotenv
load_dotenv()

from langchain_google_genai import ChatGoogleGenerativeAI
from langchain.prompts import ChatPromptTemplate
from langchain.schema.runnable import RunnableParallel, RunnableLambda
from langchain.schema.output_parser import StrOutputParser
```

```python
llm = ChatGoogleGenerativeAI(
    model="gemini-2.5-flash",
    temperature=0.3,
)
```

```python
# Create prompt templates
summarization_template = ChatPromptTemplate.from_messages([
    ("system", "You are a movie critic."),
    ("human", "Provide a brief summary of the movie {movie_name}."),
])

plot_template = ChatPromptTemplate.from_messages([
    ("system", "You are a movie critic."),
    ("human", "Analyze the plot of {movie_name}. What are its strengths and weaknesses?"),
])

character_template = ChatPromptTemplate.from_messages([
    ("system", "You are a movie critic."),
    ("human", "Analyze the characters in {movie_name}. What are their strengths and weaknesses?"),
])
```

```python
# Create individual chains
summary_chain = summarization_template | llm | StrOutputParser()
plot_analysis_chain = plot_template | llm | StrOutputParser()
character_analysis_chain = character_template | llm | StrOutputParser()

# Create parallel chain
chain = RunnableParallel({
    "summary": summary_chain,
    "plot_analysis": plot_analysis_chain,
    "character_analysis": character_analysis_chain
})
```

```python
response = chain.invoke({
    'movie_name': 'happy gilmore'
})
```

```python
print(f"===== Summary =====\n{response['summary']}\n")
print(f"===== Plot Analysis =====\n{response['plot_analysis']}\n")
print(f"===== Character Analysis =====\n{response['character_analysis']}\n")
```
