# dependencies

```python
from dotenv import load_dotenv
load_dotenv()

from langchain_google_genai import ChatGoogleGenerativeAI
from langchain.prompts import ChatPromptTemplate
from langchain.schema.runnable import RunnableLambda, RunnableSequence
```

# llm initialization

```python
llm = ChatGoogleGenerativeAI(
    model="gemini-2.5-flash",
    temperature=0.3,
)
```

# prompt template

```python
prompt_template = ChatPromptTemplate.from_messages(
     [
        ("system", "You love facts and you tell facts about {animal}"),
        ("human", "Tell me {count} facts."),
    ]
)
```

# runnables (steps in chain)

```python
format_prompt = RunnableLambda(lambda x: prompt_template.format_messages(**x))
invoke_llm = RunnableLambda(lambda x: llm.invoke(x))
parse_response = RunnableLambda(lambda x: x.content)
```

# create chain

```python
chain = RunnableSequence(
    first=format_prompt,
    middle=[invoke_llm],
    last=parse_response
)
```

# run the chain

```python
response = chain.invoke({
    "animal": "cats",
    "count": 2
})

print(response)
```
