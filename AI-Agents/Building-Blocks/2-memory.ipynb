#### Memory: Stores and retrieves relevant information across interactions. This component maintains conversation history and context to enable coherent multi-turn interactions.


#### NOTE: i am using github openai for this example.

```python
import os
from dotenv import load_dotenv
load_dotenv()

from openai import OpenAI
```

```python
TOKEN = os.getenv('GITHUB_TOKEN')
ENDPOINT = os.getenv('GITHUB_ENDPOINT')
MODEL = os.getenv('GITHUB_MODEL_NAME')

client = OpenAI(
    base_url=ENDPOINT,
    api_key=TOKEN,
)
```

```python
def ask_joke_without_memory() -> str:
    response = client.chat.completions.create(
        messages=[
            {"role": "system", "content": "You are a helpful assistant."},
            {"role": "user", "content": "Tell me a joke about programming"}
        ],
        model=MODEL
    )
    
    return response.choices[0].message.content

def ask_followup_without_memory() -> str:
    response = client.chat.completions.create(
        messages=[
            {"role": "system", "content": "You are a helpful assistant."},
            {"role": "user", "content": "What was my previous question?"}
        ],
        model=MODEL
    )
    
    return response.choices[0].message.content

def ask_joke_with_memory(joke_response: str) -> str:
    response = client.chat.completions.create(
        messages=[
            {"role": "system", "content": "You are a helpful assistant."},
            {"role": "user", "content": "Tell me a joke about programming"},
            {"role": "assistant", "content": joke_response},
            {"role": "user", "content": "What was my previous question?"}
        ],
        model=MODEL
    )
    
    return response.choices[0].message.content
```

```python
# First: Ask for a joke
print("===== Asking for a joke without memory: =====")
joke_response = ask_joke_without_memory()
print(joke_response, "\n")

# Second: Ask follow-up without memory (AI will be confused)
print("===== Asking follow-up without memory: =====")
confused_response = ask_followup_without_memory()
print(confused_response, "\n")

# Third: Ask for a joke again, but this time with the previous joke as context
print("===== Asking for a joke again with previous joke as context: =====")
joke_with_context_response = ask_joke_with_memory(joke_response)
print(joke_with_context_response, "\n")
```
